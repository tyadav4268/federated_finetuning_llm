```python
from datasets import load_dataset

def load_medqa_split(dataset_id="bigbio/med_qa", config="med_qa_en_source", split="validation"):
    """
    Loads MedQA dataset split.
    Returns a HuggingFace Dataset object.
    """
    ds = load_dataset(dataset_id, config)
    return ds[split]
val_ds = load_medqa_split()
print(val_ds[0])
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

def load_base_model(model_id):
    tokenizer = AutoTokenizer.from_pretrained(model_id, use_fast=False)
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        torch_dtype=torch.float16,
        device_map="auto",
        trust_remote_code=True
    )
    model.eval()

    return model, tokenizer
```

```python

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import PeftModel


# -----------------------------------------------------------
# Helpers
# -----------------------------------------------------------
def load_base_model(model_id: str):
    """Load base CausalLM model + tokenizer."""
    tokenizer = AutoTokenizer.from_pretrained(model_id, use_fast=False)
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        torch_dtype=torch.float16,
        device_map="auto",
        trust_remote_code=True
    )
    model.eval()
    device = next(model.parameters()).device

    return model, tokenizer, device


def load_finetuned_model(model_id: str, lora_path: str):
    """Load base model and apply LoRA weights."""
    tokenizer = AutoTokenizer.from_pretrained(model_id, use_fast=False)
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    base_model = AutoModelForCausalLM.from_pretrained(
        model_id,
        torch_dtype=torch.float16,
        device_map="auto",
        trust_remote_code=True
    )
    model = PeftModel.from_pretrained(base_model, lora_path)
    model.eval()
    device = next(model.parameters()).device

    return model, tokenizer, device


# -----------------------------------------------------------
# Prediction function
# -----------------------------------------------------------
def generate_answer(
    model,
    tokenizer,
    device,
    prompt: str,
    max_new_tokens: int = 50,
    temperature: float = 0.0
):
    """Return ONLY the raw generated text."""
    inputs = tokenizer(prompt, return_tensors="pt").to(device)

    out = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=(temperature > 0),
        temperature=temperature if temperature > 0 else None,
        pad_token_id=tokenizer.pad_token_id,
        eos_token_id=tokenizer.eos_token_id,
    )

    text = tokenizer.decode(out[0], skip_special_tokens=True)

    # Remove prompt from front
    if text.startswith(prompt):
        text = text[len(prompt):].strip()

    return text


# -----------------------------------------------------------
# Convenience functions
# -----------------------------------------------------------
def predict_base(model_id: str, prompt: str, max_new_tokens=50, temperature=0.0):
    model, tokenizer, device = load_base_model(model_id)
    return generate_answer(model, tokenizer, device, prompt, max_new_tokens, temperature)


def predict_finetuned(model_id: str, lora_path: str, prompt: str, max_new_tokens=50, temperature=0.0):
    model, tokenizer, device = load_finetuned_model(model_id, lora_path)
    return generate_answer(model, tokenizer, device, prompt, max_new_tokens, temperature)

```

```python
def build_mcq_prompt(question, options):
    opts = "\n".join([f"{o['key']}) {o['value']}" for o in options])
    return f"Question: {question}\n{opts}\nAnswer:"
```

```python
def predict_single(ex):
    prompt = build_mcq_prompt(ex["question"], ex["options"])

    base_pred = predict_base("mistralai/Mistral-7B-Instruct-v0.3", prompt)
    ft_pred   = predict_finetuned(
                "mistralai/Mistral-7B-Instruct-v0.3",
                "qlora_medqa",
                prompt)

    return {
        "prompt": prompt,
        "gold_letter": ex["answer_idx"],
        "gold_text": ex["answer"],
        "base_pred": base_pred,
        "ft_pred": ft_pred,
    }
```

```python
# --- MAIN LOGIC ---

# 1. Calculate the length of the 'question' for every sample
question_lengths = np.array([len(sample['question']) for sample in val_ds])

# 2. Find the indices that would sort the array (ascending order)
# argsort returns the indices that sort the array. The first 10 indices
# correspond to the 10 shortest questions.
sorted_indices = np.argsort(question_lengths)

# 3. Select the indices of the top 10 shortest questions
shortest_10_indices = sorted_indices[:10]

print("-" * 50)
print(f"Total samples processed: {len(val_ds)}")
print(f"Top 10 Shortest Question Indices: {shortest_10_indices.tolist()}")
print("-" * 50)
print("\nProcessing the 10 Samples with the Least Question Length:\n")

# 4. Process each of the 10 shortest samples
for i, idx_np in enumerate(shortest_10_indices):
    idx = int(idx_np)
    sample = val_ds[idx]
    
    print(f"--- Sample {i+1}/10 (Index: {idx}, Length: {question_lengths[idx]}) ---")
    print(f"Question: {sample['question']}")
    
    # Run the prediction function
    res = predict_single(sample)
    
    # Print the prediction results
    print("\n[Prediction Results]")
    for key, value in res.items():
        # Ensure that the question key is not printed again in the prediction results section
        if key != 'question':
            print(f"{key}: {value}")
    
    print("-" * 50)

print("\nProcessing complete.")
```
